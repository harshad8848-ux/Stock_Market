```python
import matplotlib.pyplot as plt
import csv

x = []
y = []

with open("stock_market.csv", 'r') as csvfile:
    lines = csv.reader(csvfile, delimiter=",")
    # Skip the header row to avoid ValueError when converting to float
    next(lines, None)

    for row in lines:
        x.append(row[0])
        y.append(float(row[1]))

print(x)
print(y)

plt.plot(x, y, color="darkblue", linestyle="-",
         marker="*", ms=15, mfc='g', mec='pink',
         label='closing price')

plt.xticks(rotation=45)

plt.xlabel('Data', fontsize=14, color='brown')
plt.ylabel('Closing price', fontsize=14, color='brown')

plt.title("Stock Market Closing Price Trend", fontsize=20)

plt.grid()
plt.legend()
plt.show()
```
