```python
import matplotlib.pyplot as plt
import csv

x = []
y = []

with open("stock_market.csv", 'r') as csvfile:
    lines = csv.reader(csvfile, delimiter=",")
    # Skip the header row to avoid ValueError when converting to float
    next(lines, None)

    for row in lines:
        x.append(row[0])
        y.append(float(row[1]))

print(x)
print(y)

plt.plot(x, y, color="darkblue", linestyle="-",
         marker="*", ms=15, mfc='g', mec='pink',
         label='closing price')

plt.xticks(rotation=45)

plt.xlabel('Data', fontsize=14, color='brown')
plt.ylabel('Closing price', fontsize=14, color='brown')

plt.title("Stock Market Closing Price Trend", fontsize=20)

plt.grid()
plt.legend()
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

rng = np.random.default_rng(42)

cc = rng.integers(800, 3000, 50)

df = pd.DataFrame({
    "Engine Capacity (CC)": cc,
    "Fuel Efficiency (km/l)": 28 - cc/150 + rng.normal(0, 1.5, 50)
})

df.loc[[3, 17], "Fuel Efficiency (km/l)"] = np.nan

df = df.dropna()

sns.set_style("whitegrid")

sns.scatterplot(
    data=df,
    x="Engine Capacity (CC)",
    y="Fuel Efficiency (km/l)",
    s=70,
    color="green",
    edgecolor="black"
)

plt.title("Engine Capacity vs Fuel Efficiency")
plt.xlabel("Engine Capacity (CC)")
plt.ylabel("Fuel Efficiency (km/l)")

plt.tight_layout()

plt.savefig("scatter_plot.png", dpi=300)

plt.show()
```
