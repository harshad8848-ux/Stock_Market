```python
import matplotlib.pyplot as plt
import csv

x = []
y = []

with open("stock_market.csv", 'r') as csvfile:
    lines = csv.reader(csvfile, delimiter=",")
    # Skip the header row to avoid ValueError when converting to float
    next(lines, None)

    for row in lines:
        x.append(row[0])
        y.append(float(row[1]))

print(x)
print(y)

plt.plot(x, y, color="darkblue", linestyle="-",
         marker="*", ms=15, mfc='g', mec='pink',
         label='closing price')

plt.xticks(rotation=45)

plt.xlabel('Data', fontsize=14, color='brown')
plt.ylabel('Closing price', fontsize=14, color='brown')

plt.title("Stock Market Closing Price Trend", fontsize=20)

plt.grid()
plt.legend()
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

rng = np.random.default_rng(42)

cc = rng.integers(800, 3000, 50)

df = pd.DataFrame({
    "Engine Capacity (CC)": cc,
    "Fuel Efficiency (km/l)": 28 - cc/150 + rng.normal(0, 1.5, 50)
})

df.loc[[3, 17], "Fuel Efficiency (km/l)"] = np.nan

df = df.dropna()

sns.set_style("whitegrid")

sns.scatterplot(
    data=df,
    x="Engine Capacity (CC)",
    y="Fuel Efficiency (km/l)",
    s=70,
    color="green",
    edgecolor="black"
)

plt.title("Engine Capacity vs Fuel Efficiency")
plt.xlabel("Engine Capacity (CC)")
plt.ylabel("Fuel Efficiency (km/l)")

plt.tight_layout()

plt.savefig("scatter_plot.png", dpi=300)

plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

months = pd.date_range("01-01-2023", periods=36, freq="MS")

t = np.arange(36)

admitted = (500 + 2*t + 60*np.sin(2*np.pi*t/12)).round()

emergency = (120 + 25*np.cos(2*np.pi*t/12)).round()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(months, admitted, marker="o", color="red")
axes[0].set_title("Monthly Patient Admission")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Patients Admitted")

axes[1].bar(months, emergency, width=20, color="green")
axes[1].set_title("Emergency Cases by Month")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Emergency Cases")

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

np.random.seed(42)

categories = ["Electronics", "Clothing", "Furniture", "Books"]

data = pd.DataFrame({
    "Product ID": range(1, 101),
    "Product Category": np.repeat(categories, 25),
    "Product Name": ["Product " + str(i) for i in range(1, 101)],
    "Brand": np.random.choice(
        ["Samsung", "Nike", "IKEA", "Sony"], 100
    ),
    "Price": np.random.randint(100, 10000, 100),
    "Customer Rating": np.round(
        np.random.uniform(2.5, 5, 100), 1
    )
})

data.to_csv("product.csv", index=False)

data = pd.read_csv("product.csv")

data.fillna(
    data.median(numeric_only=True),
    inplace=True
)

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(
    nrows=1, ncols=2, figsize=(12, 5)
)

sns.boxplot(
    data=data,
    x="Product Category",
    y="Price",
    palette=["red", "blue", "green", "orange"],
    ax=axes[0],
    hue="Product Category",
    legend=False
)

axes[0].set_title("Product Price Distribution (Boxplot)")
axes[0].set_xlabel("Product Category")
axes[0].set_ylabel("Price")

sns.violinplot(
    data=data,
    x="Product Category",
    y="Price",
    palette=["red", "blue", "green", "orange"],
    ax=axes[1],
    hue="Product Category",
    legend=False
)

axes[1].set_title("Product Price Distribution (Violin Plot)")
axes[1].set_xlabel("Product Category")
axes[1].set_ylabel("Price")

fig.suptitle("Product Price Analysis")

plt.tight_layout()
plt.savefig("product_analysis.png")
plt.show()

print(data["Price"].describe())

print(
    data.groupby("Product Category")["Price"].median()
)
```

```python

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = pd.read_csv("Electricity.csv")

sns.lineplot(
    x="Date",
    y="Electricity Consumption",
    hue="Consumer Category",
    linestyle="--",
    style="Consumer Category",
    marker="o",
    data=data
)

plt.xlabel("Time lines")
plt.ylabel("Electricity Consumption (kWh)")
plt.title("Monthly Electricity Consumption Trend")

plt.xticks(
    rotation=45,
    ha="center",
    fontsize=12,
    fontfamily="serif",
    fontweight="bold"
)

plt.yticks(
    rotation=45,
    ha="center",
    fontsize=12,
    fontfamily="serif",
    fontweight="bold"
)

plt.grid(True)
plt.show()
```

```python

import pandas as pd
import plotly.express as px

df = pd.read_csv("website_traffic.csv")

df["Date"] = pd.to_datetime(df["Date"])

fig = px.line(
    df,
    x="Date",
    y="Visitors",
    title="Website Traffic Analysis",
    hover_data=["Visitors"]
)

fig.update_traces(
    mode="lines+markers",
    line=dict(color="blue", width=4),
    marker=dict(color="red", size=10)
)

fig.update_xaxes(tickangle=45)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Number of Visitors",
    template="plotly_white"
)

fig.write_html("website.png")

fig.show()
```
